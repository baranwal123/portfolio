#  <font color="blue"> 1️⃣7️⃣  **CHECK OUTLIERS**  </font>


```python
plt.figure(figsize=(8,4))

sns.boxplot(x=df["unit_sales"])

plt.title("Outlier Detection in Unit Sales")

plt.show()

show_interpretation(
    "Unit Sales Outlier Detection",
    "A box plot is used to visualize the distribution of `unit_sales` and identify any potential outliers. This helps in understanding the spread of sales data and extreme values.",
    "#e0e7ff"
)
```

```python
# Calculate Outliers
Q1 = df["unit_sales"].quantile(0.25)
Q3 = df["unit_sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["unit_sales"] < lower_bound) |
    (df["unit_sales"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

#Number of outliers: 41



```python
# Showing Outliers using Scatter plot
plt.figure(figsize=(12,5))

plt.scatter(
    df.index,
    df["unit_sales"],
    alpha=0.6,
    label="Normal Data"
)

plt.scatter(
    outliers.index,
    outliers["unit_sales"],
    color="red",
    label="Outliers"
)

plt.title("Unit Sales with Outliers Highlighted")
plt.xlabel("Date")
plt.ylabel("Unit Sales")

plt.legend()

plt.show()
```

#  <font color="blue"> 1️⃣8️⃣ **HANDLE OUTLIERS** </font>


```python
cap_value = df["unit_sales"].quantile(0.99)

df["unit_sales_capped"] = np.where(
    df["unit_sales"] > cap_value,
    cap_value,
    df["unit_sales"]
)

show_interpretation(
    "Outlier Treatment",
    "Extreme sales values are capped at the 99th percentile to reduce their impact and improve model stability.",
    "#e0e7ff"
)

df.head()
```

```python
def adf_test(series):

    result = adfuller(series)

    print("ADF Statistic:", result[0])
    print("P-value:", result[1])

    if result[1] <= 0.05:
        print("Result: Data is Stationary")
    else:
        print("Result: Data is Not Stationary")

adf_test(df["unit_sales_capped"])

show_interpretation(
    "ADF Test for Stationarity",
    "The Augmented Dickey-Fuller (ADF) test is performed on the capped unit sales data to check for stationarity. A low p-value (typically <= 0.05) indicates that the data is stationary, a crucial assumption for many time series models.",
    "#e0e7ff"
)
```

# <font color="blue"> 3️⃣4️⃣ **MODEL EVALUATION** </font>


```python
def evaluate_model(actual, predicted, model_name):

    mae = mean_absolute_error(actual, predicted)

    rmse = np.sqrt(mean_squared_error(actual, predicted))

    mape = np.mean(
        np.abs((actual - predicted) / np.where(actual == 0, 1, actual))
    ) * 100

    print(model_name)
    print("MAE:", mae)
    print("RMSE:", rmse)
    print("MAPE:", mape)
    print("-" * 40)

evaluate_model(test, arima_prediction, "ARIMA Model")

evaluate_model(test, sarima_prediction, "SARIMA Model")

show_interpretation(
    "ARIMA and SARIMA Model Evaluation Metrics",
    "This section evaluates both the ARIMA and SARIMA models using key metrics: Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and Mean Absolute Percentage Error (MAPE). These metrics quantify the accuracy of each model's predictions against actual sales data.",
    "#e0e7ff"
)
```

```python
#@title FINAL 3D PREMIUM UI - REAL MODEL METRICS
%%writefile app.py

import streamlit as st
import joblib
import pandas as pd
import matplotlib.pyplot as plt

st.set_page_config(
    page_title="Electronic Item (Cooler) Sales Demand Forecasting System",
    page_icon="❄️",
    layout="wide"
)

css = """
<style>

[data-testid="stAppViewContainer"] {
background-image:
radial-gradient(circle at top left, rgba(250,204,21,0.35), transparent 35%),
radial-gradient(circle at bottom right, rgba(37,99,235,0.35), transparent 35%),
linear-gradient(120deg, rgba(2,6,23,0.88), rgba(30,64,175,0.76), rgba(250,204,21,0.18)),
url("https://images.unsplash.com/photo-1551288049-bebda4e38f71");
background-size: cover;
background-position: center;
background-attachment: fixed;
}

[data-testid="stHeader"] {
background: rgba(0,0,0,0);
}

.block-container {
background: rgba(255,255,255,0.91);
padding: 2.4rem;
border-radius: 34px;
box-shadow:
0px 25px 60px rgba(0,0,0,0.42),
inset 0px 0px 18px rgba(255,255,255,0.65);
border: 2px solid rgba(250,204,21,0.75);
backdrop-filter: blur(14px);
animation: fadeIn 0.9s ease-in-out;
}

@keyframes fadeIn {
from {
opacity: 0;
transform: translateY(20px);
}
to {
opacity: 1;
transform: translateY(0px);
}
}

.main-title {
text-align: center;
font-size: 50px;
font-weight: 950;
color: #0f172a;
text-shadow:
3px 3px 0px #facc15,
6px 6px 18px rgba(37,99,235,0.55),
0px 0px 25px rgba(250,204,21,0.65);
letter-spacing: 1px;
padding: 15px;
border-radius: 25px;
background: linear-gradient(135deg, rgba(255,255,255,0.50), rgba(239,246,255,0.75));
box-shadow:
inset 3px 3px 8px rgba(255,255,255,0.8),
inset -4px -4px 10px rgba(15,23,42,0.10);
}

.sub-title {
text-align: center;
font-size: 22px;
font-weight: 800;
color: #1e3a8a;
margin-bottom: 28px;
text-shadow: 1px 1px 0px rgba(250,204,21,0.45);
}

.card {
background:
linear-gradient(145deg, rgba(250,204,21,0.98), rgba(37,99,235,0.96));
color: white;
padding: 26px;
border-radius: 26px;
text-align: center;
font-size: 21px;
font-weight: 900;
box-shadow:
10px 12px 25px rgba(15,23,42,0.38),
-6px -6px 16px rgba(255,255,255,0.75),
inset 2px 2px 6px rgba(255,255,255,0.45),
inset -3px -3px 8px rgba(15,23,42,0.22);
border: 2px solid rgba(255,255,255,0.78);
transition: all 0.35s ease-in-out;
position: relative;
overflow: hidden;
}

.card::before {
content: "";
position: absolute;
top: -60%;
left: -60%;
width: 120%;
height: 120%;
background: linear-gradient(120deg, transparent, rgba(255,255,255,0.35), transparent);
transform: rotate(25deg);
transition: all 0.6s;
}

.card:hover::before {
left: 90%;
}

.card:hover {
transform: translateY(-10px) scale(1.04);
box-shadow:
16px 18px 36px rgba(15,23,42,0.50),
0px 0px 32px rgba(250,204,21,0.95);
}

.insight-box {
background: linear-gradient(145deg, rgba(239,246,255,0.96), rgba(254,249,195,0.96));
padding: 20px;
border-radius: 22px;
border-left: 8px solid #2563eb;
box-shadow:
8px 8px 18px rgba(15,23,42,0.18),
-4px -4px 12px rgba(255,255,255,0.8),
inset 1px 1px 6px rgba(255,255,255,0.7);
font-size: 17px;
font-weight: 650;
color: #0f172a;
margin-bottom: 16px;
transition: all 0.25s ease-in-out;
}

.insight-box:hover {
transform: translateY(-5px);
box-shadow:
12px 14px 28px rgba(15,23,42,0.25),
0px 0px 18px rgba(37,99,235,0.25);
}

.workflow-step {
background: linear-gradient(135deg, #0f172a, #1e3a8a, #2563eb);
color: white;
padding: 19px;
border-radius: 20px;
font-weight: 850;
text-align: center;
box-shadow:
8px 10px 22px rgba(37,99,235,0.38),
inset 2px 2px 8px rgba(255,255,255,0.22);
margin-bottom: 14px;
border: 1.5px solid rgba(255,255,255,0.35);
transition: 0.3s;
}

.workflow-step:hover {
transform: scale(1.035) translateY(-4px);
box-shadow: 0px 0px 26px rgba(37,99,235,0.65);
}

h1, h2, h3 {
color: #1e3a8a;
font-weight: 900;
text-shadow: 1px 1px 0px rgba(250,204,21,0.35);
}

.stButton > button {
background: linear-gradient(135deg, #facc15, #f97316);
color: #0f172a;
font-size: 18px;
font-weight: 900;
border-radius: 18px;
padding: 0.8rem 1.6rem;
border: none;
box-shadow:
0px 10px 22px rgba(249,115,22,0.45),
inset 2px 2px 5px rgba(255,255,255,0.55);
transition: all 0.25s ease-in-out;
}

.stButton > button:hover {
background: linear-gradient(135deg, #2563eb, #1e40af);
color: white;
transform: translateY(-5px) scale(1.03);
box-shadow: 0px 14px 28px rgba(37,99,235,0.60);
}

.stDownloadButton > button {
background: linear-gradient(135deg, #2563eb, #1d4ed8);
color: white;
font-size: 17px;
font-weight: 900;
border-radius: 18px;
padding: 0.8rem 1.5rem;
border: none;
box-shadow:
0px 10px 22px rgba(37,99,235,0.45),
inset 2px 2px 5px rgba(255,255,255,0.35);
transition: all 0.25s ease-in-out;
}

.stDownloadButton > button:hover {
background: linear-gradient(135deg, #facc15, #f97316);
color: #0f172a;
transform: translateY(-5px);
}

[data-testid="stSidebar"] {
background:
linear-gradient(180deg, rgba(2,6,23,0.98), rgba(30,58,138,0.96), rgba(250,204,21,0.88));
box-shadow: 8px 0px 25px rgba(0,0,0,0.35);
}

[data-testid="stSidebar"] * {
color: white;
font-weight: 650;
}

[data-testid="stMetric"] {
background: linear-gradient(145deg, rgba(239,246,255,0.98), rgba(254,249,195,0.98));
padding: 19px;
border-radius: 22px;
box-shadow:
8px 10px 20px rgba(15,23,42,0.22),
-4px -4px 12px rgba(255,255,255,0.75),
inset 2px 2px 8px rgba(255,255,255,0.55);
border-left: 7px solid #facc15;
transition: all 0.25s ease-in-out;
}

[data-testid="stMetric"]:hover {
transform: translateY(-5px) scale(1.02);
box-shadow: 0px 0px 25px rgba(250,204,21,0.55);
}

[data-testid="stFileUploader"] {
background: linear-gradient(145deg, #eff6ff, #fefce8);
padding: 18px;
border-radius: 22px;
border: 2px dashed #2563eb;
box-shadow: inset 2px 2px 10px rgba(37,99,235,0.12);
}

div[data-testid="stDataFrame"] {
border-radius: 18px;
box-shadow: 0px 10px 25px rgba(15,23,42,0.18);
}

.footer {
text-align: center;
color: #1e3a8a;
font-weight: 950;
font-size: 21px;
text-shadow:
1px 1px 0px #facc15,
0px 0px 18px rgba(250,204,21,0.65);
padding: 24px;
border-radius: 24px;
background: linear-gradient(145deg, rgba(239,246,255,0.95), rgba(254,249,195,0.95));
box-shadow:
8px 8px 20px rgba(15,23,42,0.20),
inset 2px 2px 8px rgba(255,255,255,0.65);
}

hr {
border: none;
height: 3px;
background: linear-gradient(90deg, transparent, #2563eb, #facc15, #2563eb, transparent);
border-radius: 10px;
margin: 28px 0px;
}

</style>
"""
st.markdown(css, unsafe_allow_html=True)

# ==============================
# LOAD MODEL
# ==============================
try:
    model = joblib.load("best_timeseries_model.pkl")
    best_model_name = joblib.load("best_model_name.pkl")
except:
    st.error("Model files not found. Please upload best_timeseries_model.pkl and best_model_name.pkl.")
    st.stop()

# ==============================
# FIXED PROJECT VALUES
# ==============================
PROJECT_NAME = "Electronic Item (Cooler) Demand Forecasting System"
STORE_NUMBER = 25
ITEM_NUMBER = 305229
LAST_DATASET_DATE = pd.to_datetime("2017-08-15")

# ==============================
# REAL MODEL METRICS FROM YOUR OUTPUT
# ==============================
ARIMA_MAE = 7.555064
ARIMA_RMSE = 10.496602
ARIMA_MAPE = 76.526256

SARIMA_MAE = 7.164537
SARIMA_RMSE = 9.812843
SARIMA_MAPE = 76.811868

ERROR_REDUCTION = ((ARIMA_RMSE - SARIMA_RMSE) / ARIMA_RMSE) * 100

metrics_df = pd.DataFrame({
    "Model": ["ARIMA", "SARIMA"],
    "MAE": [ARIMA_MAE, SARIMA_MAE],
    "RMSE": [ARIMA_RMSE, SARIMA_RMSE],
    "MAPE": [ARIMA_MAPE, SARIMA_MAPE]
})

# ==============================
# SIDEBAR
# ==============================
st.sidebar.title("📌 Project Menu")
st.sidebar.success(f"Best Model: {best_model_name}")

st.sidebar.header("Project Name")
st.sidebar.write(PROJECT_NAME)

st.sidebar.header("Business Objective")
st.sidebar.write("Predict cooler demand for Store 25 and Item 305229.")
st.sidebar.write("Reduce stockouts, overstocking, and inventory cost.")
st.sidebar.write("Improve stock replenishment decisions.")

st.sidebar.header("Selected Scope")
st.sidebar.write(f"Store Number: {STORE_NUMBER}")
st.sidebar.write(f"Item Number: {ITEM_NUMBER}")
st.sidebar.write("Target Variable: unit_sales")

st.sidebar.header("Model Meaning")
st.sidebar.write("ARIMA → Non-seasonal forecasting")
st.sidebar.write("SARIMA → Seasonal forecasting")
st.sidebar.write("RMSE → Model selection metric")

# ==============================
# MAIN TITLE
# ==============================
st.markdown(
    f"<div class='main-title'>❄️ {PROJECT_NAME}</div>",
    unsafe_allow_html=True
)

st.markdown(
    f"<div class='sub-title'>Store {STORE_NUMBER} | Item {ITEM_NUMBER} | ARIMA vs SARIMA</div>",
    unsafe_allow_html=True
)

st.write("---")

st.info(f"""
This dashboard predicts future demand for an electronic item cooler.
It focuses on Store {STORE_NUMBER} and Item {ITEM_NUMBER}.
Forecasting starts from the dataset's last available date, not from the current system date.
""")

# ==============================
# TOP CARDS
# ==============================
c1, c2, c3, c4 = st.columns(4)

with c1:
    st.markdown(f"<div class='card'>🏬 Store<br>{STORE_NUMBER}</div>", unsafe_allow_html=True)

with c2:
    st.markdown(f"<div class='card'>❄️ Cooler Item<br>{ITEM_NUMBER}</div>", unsafe_allow_html=True)

with c3:
    st.markdown(f"<div class='card'>✅ Best Model<br>{best_model_name}</div>", unsafe_allow_html=True)

with c4:
    st.markdown("<div class='card'>📉 Metric<br>Lowest RMSE</div>", unsafe_allow_html=True)

st.write("---")

# ==============================
# FORECAST SETTINGS
# ==============================
st.header("🔮 Generate Cooler Demand Forecast")

st.info(f"""
This section predicts future daily cooler demand for Store {STORE_NUMBER} and Item {ITEM_NUMBER}.
Forecast dates start from {(LAST_DATASET_DATE + pd.Timedelta(days=1)).date()}.
""")

col1, col2, col3 = st.columns(3)

with col1:
    steps = st.slider("Select number of future days", 1, 365, 30)

with col2:
    chart_type = st.selectbox("Choose chart type", ["Line Chart", "Area Chart", "Bar Chart"])

with col3:
    show_table = st.radio("Show forecast table?", ["Yes", "No"], horizontal=True)

if st.button("🚀 Generate Demand Forecast"):

    forecast = model.forecast(steps=steps)

    future_dates = pd.date_range(
        start=LAST_DATASET_DATE + pd.Timedelta(days=1),
        periods=steps,
        freq="D"
    )

    forecast_df = pd.DataFrame({
        "Forecast_Date": future_dates,
        "Future_Day": range(1, steps + 1),
        "Forecast_Unit_Sales": forecast
    })

    st.success("Cooler demand forecast generated successfully!")

    st.subheader("📌 Forecast Summary")

    first_value = forecast_df["Forecast_Unit_Sales"].iloc[0]
    last_value = forecast_df["Forecast_Unit_Sales"].iloc[-1]

    if last_value > first_value:
        trend = "Increasing Demand 📈"
        recommendation = "Increase cooler stock quantity to avoid stockouts."
    elif last_value < first_value:
        trend = "Decreasing Demand 📉"
        recommendation = "Maintain controlled cooler inventory to avoid overstocking."
    else:
        trend = "Stable Demand ➖"
        recommendation = "Maintain normal cooler inventory level."

    s1, s2, s3, s4, s5 = st.columns(5)

    with s1:
        st.metric("Future Days", steps)

    with s2:
        st.metric("Average Demand", round(forecast_df["Forecast_Unit_Sales"].mean(), 2))

    with s3:
        st.metric("Highest Demand", round(forecast_df["Forecast_Unit_Sales"].max(), 2))

    with s4:
        st.metric("Lowest Demand", round(forecast_df["Forecast_Unit_Sales"].min(), 2))

    with s5:
        st.metric("Demand Trend", trend)

    st.info(f"""
    Forecast Trend: {trend}

    Business Recommendation: {recommendation}
    """)

    st.subheader("📈 Forecast Visualization")

    chart_data = forecast_df.set_index("Forecast_Date")["Forecast_Unit_Sales"]

    if chart_type == "Line Chart":
        st.line_chart(chart_data)
    elif chart_type == "Area Chart":
        st.area_chart(chart_data)
    else:
        st.bar_chart(chart_data)

    if show_table == "Yes":
        st.subheader("📋 Forecast Table")
        st.dataframe(forecast_df, use_container_width=True)

    st.subheader("🔍 Forecast Confidence Interval")

    try:
        forecast_result = model.get_forecast(steps=steps)
        confidence_interval = forecast_result.conf_int()
        confidence_interval = confidence_interval.reset_index(drop=True)
        confidence_interval["Forecast_Date"] = future_dates
        confidence_interval["Future_Day"] = range(1, steps + 1)

        st.dataframe(confidence_interval, use_container_width=True)

    except:
        st.warning("Confidence interval is not available for this saved model.")

    st.subheader("📊 Detailed Forecast Graph")

    fig, ax = plt.subplots(figsize=(12, 5))
    ax.plot(
        forecast_df["Forecast_Date"],
        forecast_df["Forecast_Unit_Sales"],
        marker="o",
        linewidth=2
    )

    ax.set_title(f"{best_model_name} Cooler Demand Forecast | Store {STORE_NUMBER} | Item {ITEM_NUMBER}")
    ax.set_xlabel("Forecast Date")
    ax.set_ylabel("Forecast Unit Sales")
    ax.grid(True)

    plt.xticks(rotation=45)
    st.pyplot(fig)

    csv = forecast_df.to_csv(index=False).encode("utf-8")

    st.download_button(
        label="⬇️ Download Demand Forecast CSV",
        data=csv,
        file_name=f"cooler_store_{STORE_NUMBER}_item_{ITEM_NUMBER}_forecast.csv",
        mime="text/csv"
    )

# ==============================
# DATASET UPLOAD
# ==============================
st.write("---")
st.header("📂 Upload Dataset for Analysis")

uploaded_file = st.file_uploader("Upload CSV file", type=["csv"])

if uploaded_file is not None:

    df = pd.read_csv(uploaded_file)
    st.success("Dataset uploaded successfully!")

    if "store_nbr" in df.columns and "item_nbr" in df.columns:
        filtered_df = df[
            (df["store_nbr"] == STORE_NUMBER) &
            (df["item_nbr"] == ITEM_NUMBER)
        ]

        st.success(f"Filtered dataset for Store {STORE_NUMBER} and Item {ITEM_NUMBER}")

        if not filtered_df.empty:
            df = filtered_df
        else:
            st.warning("No records found for this store-item combination in uploaded dataset.")

    st.subheader("👀 Dataset Preview")
    st.dataframe(df.head(), use_container_width=True)

    st.subheader("📊 Dataset Overview")

    d1, d2, d3, d4 = st.columns(4)

    with d1:
        st.metric("Rows", df.shape[0])
    with d2:
        st.metric("Columns", df.shape[1])
    with d3:
        st.metric("Missing Values", df.isnull().sum().sum())
    with d4:
        st.metric("Duplicate Rows", df.duplicated().sum())

    st.subheader("📌 Column Names")
    st.write(df.columns.tolist())

    numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

    if len(numeric_cols) > 0:
        selected_col = st.selectbox("Select numeric column for analysis", numeric_cols)

        st.subheader(f"📈 Trend of {selected_col}")
        st.line_chart(df[selected_col])

        st.subheader("📌 Basic Statistics")
        st.dataframe(df[selected_col].describe())

        st.subheader("📉 Missing Values by Column")
        missing_df = pd.DataFrame({
            "Column": df.columns,
            "Missing_Count": df.isnull().sum().values
        })
        st.dataframe(missing_df, use_container_width=True)

# ==============================
# DATA QUALITY CHECKS
# ==============================
st.write("---")
st.header("🧹 Data Quality Checks")

q1, q2, q3, q4 = st.columns(4)

with q1:
    st.metric("Missing Values", "Handled")

with q2:
    st.metric("Outliers", "Checked")

with q3:
    st.metric("Stationarity", "ADF Tested")

with q4:
    st.metric("Seasonality", "SARIMA Used")

st.markdown("""
<div class="insight-box">
Clean and consistent data improves forecasting accuracy. Missing values, outliers,
stationarity and seasonal behavior were checked before building ARIMA and SARIMA models.
</div>
""", unsafe_allow_html=True)

# ==============================
# MODEL METRICS & COMPARISON
# ==============================
st.write("---")
st.header("📊 Model Metrics & Comparison")

m1, m2, m3, m4 = st.columns(4)

with m1:
    st.metric("ARIMA RMSE", round(ARIMA_RMSE, 2))

with m2:
    st.metric("SARIMA RMSE", round(SARIMA_RMSE, 2))

with m3:
    st.metric("Best Model", best_model_name)

with m4:
    st.metric("Error Reduction", f"{round(ERROR_REDUCTION, 2)}%")

st.subheader("📉 RMSE Comparison Graph")

fig2, ax2 = plt.subplots(figsize=(8, 4))
ax2.bar(metrics_df["Model"], metrics_df["RMSE"])
ax2.set_title("ARIMA vs SARIMA RMSE Comparison")
ax2.set_xlabel("Model")
ax2.set_ylabel("RMSE")
ax2.grid(axis="y")
st.pyplot(fig2)

st.subheader("🏆 Best Model Selection")

st.success(f"""
The final selected model is {best_model_name}.
It was selected because SARIMA RMSE ({round(SARIMA_RMSE, 2)}) is lower than ARIMA RMSE ({round(ARIMA_RMSE, 2)}).
Lower RMSE means lower forecasting error.
""")

st.subheader("📋 Model Metrics Table")
st.dataframe(metrics_df.round(2), use_container_width=True)

# ==============================
# BUSINESS INSIGHTS
# ==============================
st.write("---")
st.header("💡 Business Insights")

b1, b2 = st.columns(2)

with b1:
    st.markdown("<div class='insight-box'>✅ Forecasting helps maintain enough cooler stock before demand increases.</div>", unsafe_allow_html=True)
    st.markdown("<div class='insight-box'>📦 It reduces stockouts during high-demand periods.</div>", unsafe_allow_html=True)
    st.markdown("<div class='insight-box'>💰 It helps reduce unnecessary inventory holding cost.</div>", unsafe_allow_html=True)

with b2:
    st.markdown("<div class='insight-box'>📉 Controlled inventory reduces overstocking risk.</div>", unsafe_allow_html=True)
    st.markdown("<div class='insight-box'>🏬 Store-level forecasting improves local demand planning.</div>", unsafe_allow_html=True)
    st.markdown("<div class='insight-box'>🚀 Data-driven forecasting supports better business decisions.</div>", unsafe_allow_html=True)

# ==============================
# METHODOLOGY
# ==============================
st.write("---")
st.header("🧪 Methodology")

w1, w2, w3 = st.columns(3)

with w1:
    st.markdown("<div class='workflow-step'>1️⃣ Data Collection</div>", unsafe_allow_html=True)
    st.markdown("<div class='workflow-step'>2️⃣ Data Cleaning</div>", unsafe_allow_html=True)
    st.markdown("<div class='workflow-step'>3️⃣ Store-Item Filtering</div>", unsafe_allow_html=True)

with w2:
    st.markdown("<div class='workflow-step'>4️⃣ Time Series Preparation</div>", unsafe_allow_html=True)
    st.markdown("<div class='workflow-step'>5️⃣ Stationarity Check</div>", unsafe_allow_html=True)
    st.markdown("<div class='workflow-step'>6️⃣ ARIMA Modeling</div>", unsafe_allow_html=True)

with w3:
    st.markdown("<div class='workflow-step'>7️⃣ SARIMA Modeling</div>", unsafe_allow_html=True)
    st.markdown("<div class='workflow-step'>8️⃣ RMSE Evaluation</div>", unsafe_allow_html=True)
    st.markdown("<div class='workflow-step'>9️⃣ Streamlit Deployment</div>", unsafe_allow_html=True)

# ==============================
# MODEL EXPLANATION
# ==============================
st.write("---")
st.header("🧠 Model Explanation")

st.info("""
ARIMA captures non-seasonal demand patterns.

SARIMA captures seasonal demand patterns such as weekly or monthly fluctuations.

Since cooler demand can change due to seasonal behavior, SARIMA is useful for capturing repeated demand patterns.

Although MAPE is high, RMSE was used as the primary selection metric because RMSE gives a stable error measurement for forecasting.
""")

# ==============================
# PROJECT SUMMARY
# ==============================
st.write("---")
st.header("📌 Project Summary")

p1, p2, p3 = st.columns(3)

with p1:
    st.metric("Project", "Cooler Forecasting")

with p2:
    st.metric("Store | Item", f"{STORE_NUMBER} | {ITEM_NUMBER}")

with p3:
    st.metric("Models", "ARIMA vs SARIMA")

st.markdown(f"""
<div class="insight-box">
{PROJECT_NAME} predicts future demand for cooler sales using historical time series data.
The dashboard helps in inventory planning, stock control, demand understanding and business decision-making.
</div>
""", unsafe_allow_html=True)

# ==============================
# PROJECT TEAM
# # ==============================
# st.write("---")
# st.header("👨‍💻 Project Team")

# t1, t2, t3, t4 = st.columns(4)

# with t1:
#     st.info("Shlok Baranwal\n\nML Engineer")

# with t2:
#     st.info("Team Member 2\n\nData Analyst")

# with t3:
#     st.info("Team Member 3\n\nVisualization")

# with t4:
#     st.info("Team Member 4\n\nDeployment")

# ==============================
# FOOTER
# ==============================
st.write("---")
st.markdown(
    f"<div class='footer'>🏆 Developed for WIPRO Capstone Project<br>{PROJECT_NAME}<br>Store {STORE_NUMBER} | Item {ITEM_NUMBER} | ARIMA vs SARIMA</div>",
    unsafe_allow_html=True
)
```
